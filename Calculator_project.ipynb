```python
#My First ever python project of Calculator
#So first of all we will create a varaible in the name of NUM1 in a such a way
#to ask from the user
Num1 = float(input("Enter first number: "))
#So her is the second variable as we are creating here to again from the user
#to enter your number
Num2 = float(input("Enter second number: "))
#So in the third position we are creating the variable in the name of Operator
#to ask form the user that which operation do you want to continue
Operator = input("Enter operator (+, -, *, /): ")
## Initialize result variable to store the calculation outcome
result = None
# Now we will to create the operators and also use else if conditions
if Operator == "+":
    result = Num1 + Num2
elif Operator == "-":
    result = Num1 - Num2
elif Operator == "*":
    result = Num1 * Num2
elif Operator == "/":
    if Num2 == 0:
        print("Error: Cannot divide by zero")
    else:
        result = Num1 / Num2
else:
    print("Error: Invalid operator")

# Agar result calculate ho gaya hai toh hi print karein
if result is not None:
    print(f"Result: {result}")
```
