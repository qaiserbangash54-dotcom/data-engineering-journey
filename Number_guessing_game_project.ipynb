```python
# Now are going to do a project of Number guessing game
# So let's start from creating a variable in the name of secret Number
# first we shall import a function randomint.random

import random

secret_number = random.randint(1, 50)

print('=====Number Guessing Game(Unlimited Tries)=====')

while True:
  Guess = int(input('Enter your Guess(1-50):'))

  if Guess == secret_number:
    print(f"Congratulations! You guessed it right! The secret number was {secret_number}.")
    break

  elif Guess > secret_number:
    print('Too High! Try a lower number.')

  elif Guess < secret_number:
    print("Too low! Try a higher number.")

  else:
    print('invalid Number!please enter a number bewtween 1 and 50.')







```
